```python
import mfem.ser as mfem
import numpy as np
import matplotlib.pyplot as plt
```

```python
mesh = mfem.Mesh()
mesh.Load("../bin/radial_mesh.mesh")
x_radial = mfem.GridFunction(mesh,"../bin/radial_concentration.gf")
x_potential = mfem.GridFunction(mesh,"../bin/potential.gf")
x_diffusion = mfem.GridFunction(mesh,"../bin/diffusion_concentration.gf")
```

```python
coordinates = []
for i in range(mesh.GetNV()): # Gets number of vertices
    vertices = mesh.GetVertexArray(i) 
    coordinates.append(vertices[0])
coordinates = np.array(coordinates)
```

```python
phi_radial = []
for i in range(x_radial.Size()): # Gets number of DOF values
    phi_radial.append(x_radial[i])
phi_radial = np.array(phi_radial)

phi_potential = []
for i in range(x_potential.Size()): # Gets number of DOF values
    phi_potential.append(x_potential[i])
phi_potential = np.array(phi_potential)

phi_diffusion = []
for i in range(x_diffusion.Size()): # Gets number of DOF values
    phi_diffusion.append(x_diffusion[i])
phi_diffusion = np.array(phi_diffusion)
```

```python
plt.plot(coordinates,phi_radial,marker="o")
plt.xlabel("x")
plt.ylabel("phi(x)")
```

```python
plt.plot(coordinates,phi_diffusion,marker="o")
plt.xlabel("x")
plt.ylabel("phi(x)")
```
